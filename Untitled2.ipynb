```python
class Student:
    def __init__(self,marks):
        self.__marks=marks
    @property
    def marks(self):
        return self.__marks
student=Student(45)
print(student.marks)
```

```python
#employee management system
class Employee:
    def __init__(self,name,salary):
        self.__name=name
        self.__salary=salary
#add getter
    @property
    def salary(self):
        return self.__salary
#add setter
    @salary.setter
    def salary(self,salary):
       if salary>0:
           self.__salary=salary
       else:
           print("salary cannot be negative")
```

```python
employee = Employee("SaiKamal",30000)
print("employee salary:",employee.salary)
```

```python
employee.salary=40000
print(employee.salary)
```
